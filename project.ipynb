```python
def get_model_chars(model):
    total_size = len(model.model)
    single_son_verts = 0
    for value in model.model.values():
        if len(set(value)) == 1:
            single_son_verts += 1
    print(single_son_verts)
    return total_size, single_son_verts, single_son_verts / total_size
```

```python
def get_model(filepath, n):
    with open(filepath, encoding='utf-8') as file:
        text = file.read()

    tokens = word_tokenize(text)
    words = [token.lower() for token in tokens if token.isalpha() or token == '.']

    model = MarkovModel(n, words)
    print(get_model_chars(model))

    return model
```
