```python
from gen_text import MarkovModel

import nltk
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)
nltk.download('punkt', quiet=True)
```

```python
def get_model_chars(model):
    total_size = len(model.model)
    single_son_verts = 0
    for value in model.model.values():
        if len(set(value)) == 1:
            single_son_verts += 1
    print(single_son_verts)
    return total_size, single_son_verts, single_son_verts / total_size
```

```python
def get_model(filepath, n):
    with open(filepath) as file:
        text = file.read()

    tokens = word_tokenize(text)
    words = [token.lower() for token in tokens if token.isalpha() or token == '.']

    model = MarkovModel(n, words)
    print(get_model_chars(model))

    return model
```

```python
model_war_and_peace = get_model("war_and_peace.txt", 2)
print(" ".join(model_war_and_peace.gen_from_model(max_gen=20)))
```

```python
model_emma = get_model("emma.txt", 3)
print(" ".join(model_emma.gen_from_model(max_gen=20)))
```
